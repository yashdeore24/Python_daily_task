```python
# find potentially numeric object columns

for col in obj_clos:
    coverted = pd.to_numeric(df[col], errors = 'coerce')

    # count how many values become numbers
    num_count = coverted.notna().sum()

    if num_count > 0:
        print(f"{col} --> potentially numeric ({num_count} numarice values)")
```

# Q 18

```python
# Convert to datetime
df['Join_Date'] = pd.to_datetime(df['Join_Date'], errors='coerce')

```

```python
# Count missing dates
missing_dates = df['Join_Date'].isna().sum()
print('Missing dates after conversion:', missing_dates)

```

```python
# Fill missing dates using previous valid date
df['Join_Date'] = df['Join_Date'].fillna(method='ffill')
```

```python

# If the first rows are still missing, use backward fill
df['Join_Date'] = df['Join_Date'].fillna(method='bfill')

```

```python
# Final check
print('Missing dates after filling:', df['Join_Date'].isna().sum())

```

```python
# Display sample
print(df['Join_Date'].head(10))
```
