# 18. Task

```python
# 18. Take 5 city names as input from user and store them in a list. Then print them in reverse order.

city1 = input("Enter the name of city:")
city2 = input("Enter the name of city:")
city3 = input("Enter the name of city:")
city4 = input("Enter the name of city:")
city5 = input("Enter the name of city:")

lists = [city1, city2, city3, city4, city5]

print("Orderd list of an citys:",lists)

print("Reverse order of citys name:",lists[::-1])

lists.sort(reverse = True)

print("Decending order list of an citys:",lists)

```
