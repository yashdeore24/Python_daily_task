# Part A – Abstraction Without super() (1–15) 


## 1.Task

```python
"""
1. Class Name: Shape 
Problem 
Create an abstract class named Shape. 
Explanation 
A shape can have different formulas for area and perimeter. Therefore, create two abstract  methods: 
• area()  
• perimeter()  
Create subclasses like Circle and Rectangle and implement both methods.

"""


from abc import ABC, abstractmethod
import math

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Circle(Shape):

    def __init__(self,r):
        self.Radius = r

    def area(self):
        print(f"Area of circle is : {math.pi * self.Radius**2}")
        

    def perimeter(self):
        print(f"Perimeter of circle is : {2 * math.pi * self.Radius}")

class Rectangle(Shape):

    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        print(f"Area of Rectangle is : {self.length * self.width}")
        

    def perimeter(self):
        print(f"Perimeter of circle is : { 2 * (self.length + self.width)}") 

cir = Circle(5)
rec = Rectangle(3,5)

cir.area()
cir.perimeter()
rec.area()
rec.perimeter()
        
```

```python
"""
3. Class Name: Animal 
Problem 
Create an abstract class named Animal. 
Explanation 
Every animal makes a sound, but the sound depends on the animal.
Sumit Sir Notes  
Create an abstract method: 
• make_sound()  
Create subclasses: 
• Dog  
• Cat  
Each should implement its own sound. 

"""

# Abstract Class Animal
class Animal(ABC):

    # abstract method make sound
    @abstractmethod
    def make_sound(self):
        pass

# child class Dog
class Dog(Animal):

    def make_sound(self):
        print("Bhuuu !")

# child class Cat
class Cat(Animal):

    def make_sound(self):
        print("Mauu !")


dog = Dog()
cat = Cat()

dog.make_sound()
cat.make_sound()
    
    
    
```

## 11.Task

```python
"""1. Class Name: FileHandler 
Problem 
Create an abstract class named FileHandler. Explanation 
Files can be read and written differently. Create abstract methods: 
• read()  
• write(data)  
Create subclasses like: 
• TextFileHandler  
• CSVFileHandler  
"""

from abc import ABC, abstractmethod

# Abstract Class
class FileHandler(ABC):
    
    
        
    @abstractmethod
    def read(self):
        pass

    @abstractmethod
    def write(self,data):
        pass


# Text File Handler
class TextFileHandler(FileHandler):

    def read(self):
        print("Reading data from a Text file.")

    def write(self,data):
        print(f"Writing '{data}' to a Text file.")


# CSV File Handler
class CSVFileHandler(FileHandler):

    def read(self):
        print("Reading data from a CSV file.")

    def write(self, data):
        print(f"Writing '{data}' to a CSV file.")


# Driver Code
text = TextFileHandler()
text.read()
text.write("Hello World")

print()

csv = CSVFileHandler()
csv.read()
csv.write("Name,Age,City")
```

## 13.Task

```python
"""
13. Class Name: Calculator
Problem
Create an abstract class named Calculator.

Sumit Sir Notes
Explanation
Different calculators perform different operations.
Create abstract method:
• compute(a, b)
Create subclasses:
• AdditionCalculator
• MultiplicationCalculator

"""

class Calculator(ABC):
    def __init__(self,a,b):
        self.a = a
        self.b = b

    @abstractmethod
    def compute(self):
        pass

class AdditionCalculator(Calculator):

    def compute(self):
        print(f"Addition is : {self.a + self.b}")

class MultiplicationCalculator(Calculator):

    def compute(self):
        print(f"Multiplication is : {self.a * self.b}")


add = AdditionCalculator(2,4)
mul = MultiplicationCalculator(4,6)

add.compute()
mul.compute()
        
```

```python
"""
14. Class Name: Database
Problem
Create an abstract class named Database.
Explanation
Different databases connect and execute queries differently.
Create abstract methods:
• connect()
• query(sql)
Create subclasses:
• MySQLDatabase
• MongoDatabase
"""

from abc import ABC, abstractmethod


# Abstract Class
class Database(ABC):

    @abstractmethod
    def connect(self):
        pass

    @abstractmethod
    def query(self, sql):
        pass


# Child Class 1
class MySQLDatabase(Database):

    def connect(self):
        print("Connected to MySQL Database.")

    def query(self, sql):
        print(f"MySQL Executing Query: {sql}")


# Child Class 2
class MongoDatabase(Database):

    def connect(self):
        print("Connected to MongoDB Database.")

    def query(self, sql):
        print(f"MongoDB Executing Query: {sql}")


# Driver Code
mysql = MySQLDatabase()
mysql.connect()
mysql.query("SELECT * FROM students;")

print("-" * 40)

mongo = MongoDatabase()
mongo.connect()
mongo.query("db.students.find({})")
```

```python
"""
15. Class Name: Shape3D
Problem
Create an abstract class named Shape3D.
Explanation
Three-dimensional shapes have different formulas.
Create abstract methods:
• volume()
• surface_area()

Sumit Sir Notes
Createsubclasses:
• Cube
• Sphere
"""

import math
class Shape3D(ABC):

    @abstractmethod
    def volume(self):
        pass
    
    @abstractmethod
    def surface_area(self):
        pass

class Cube(Shape3D):

    def __init__(self,side):
        self.side=side
    
   
    def volume(self):
        print("Volume of cube is", self.side**3)
        
    def surface_area(self):
        print("Surface area of cube is ", 6 * self.side**2)

class Sphere(Shape3D):

    def __init__(self,r):
        self.r= r
    
   
    def volume(self):
        print("Volume of Shere is", (3/4 * math.pi * self.r**3))
        
    def surface_area(self):
        print("Surface area of Sphere is ", 4 *math.pi *self.r**2)

        

sp = Sphere(5)
cq = Cube(4)

sp.volume()
sp.surface_area()

cq.volume()
cq.surface_area()

```

## 28.Task

```python

from abc import ABC, abstractmethod

# Abstract Class
class Vehicle(ABC):

    def __init__(self, brand):
        self.brand = brand

    @abstractmethod
    def start(self):
        pass


# First Level Inheritance
class Car(Vehicle):

    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

    def start(self):
        print(f"{self.brand} {self.model} car is starting...")


# Second Level Inheritance
class ElectricCar(Car):

    def __init__(self, brand, model, battery):
        super().__init__(brand, model)
        self.battery = battery

    def display(self):
        print("Electric Car Details")
        print(f"Brand   : {self.brand}")
        print(f"Model   : {self.model}")
        print(f"Battery : {self.battery} kWh")



car = ElectricCar("Tesla", "Model 3", 75)

car.start()
car.display()
```

## 30.Task

```python
from abc import ABC, abstractmethod


# Abstract Class
class Report(ABC):

    def __init__(self, title):
        self.title = title

    def show_header(self):
        print("===== Report =====")
        print(f"Title : {self.title}")

    @abstractmethod
    def generate(self):
        pass


# PDF Report
class PDFReport(Report):

    def __init__(self, title):
        super().__init__(title)

    def generate(self):
        super().show_header()
        print("Generating PDF Report...")


# HTML Report
class HTMLReport(Report):

    def __init__(self, title):
        super().__init__(title)

    def generate(self):
        super().show_header()
        print("Generating HTML Report...")


# Object Creation and Method Calling
pdf = PDFReport("Student Result")
pdf.generate()

print()

html = HTMLReport("Employee Report")
html.generate()
```
