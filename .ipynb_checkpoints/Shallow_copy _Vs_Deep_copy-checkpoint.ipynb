## 1. Create a list original = [1, 2, [3, 4], 5]. Make a shallow copy named shallow and change shallow[2][0] = 99. Print both lists and explain the result.

```python
import copy

original = [1, 2, [3, 4], 5]
shallow = copy.copy(original)

shallow[2][0] = 99

print("Original list:",original)
print("Shallow:",shallow)
```

## 2. Do the same as above but use deepcopy instead of shallow copy. Show the difference.

```python
import copy

original = [1, 2, [3, 4], 5]
shallow = copy.deepcopy(original)

shallow[2][0] = 99

print("Original list:",original)
print("Shallow:",shallow)
```

## 3. Write a program to demonstrate shallow copy using copy.copy() on a nested list.

```python
import copy

original = [[1,2,3],[4,5,6],[7,8,9]]
shallow = copy.copy(original)

print(original)
print(shallow)

print(id(original))
print(id(shallow))

print(id(original[0][1]))
print(id(shallow[0][1]))
```

## 7. Write a function manual_deep_copy(lst) that manually creates a deep copy of a nested list  without using copy.deepcopy(). 


```python
def manual_deep_copy(lst):
    new_list = []

    for item in lst:
        if isinstance(item, list):
            # Recursively copy the nested list
            new_list.append(manual_deep_copy(item))
        else:
            # Copy normal elements
            new_list.append(item)

    return new_list


# Original nested list
original = [1, [2, 3], [4, [5, 6]]]

# Manual deep copy
copied = manual_deep_copy(original)
print(copied)

# Modify the copied list
copied[1].append(100)
copied[2][1].append(200)

print("Original:", original)
print("Copied  :", copied)
```

## 9. (Hard) Create a function that detects whether a copy is shallow or deep by modifying a  nested element and checking the original. 


```python
import copy

def detect_copy(original, copied):
    # Store the original nested list
    old_original = original[1][:]

    # Modify the copied nested list
    copied[1].append(100)

    # Check whether the original changed
    if original[1] != old_original:
        return "Shallow Copy"
    else:
        return "Deep Copy"


# Original list
original = [1, [2, 3], 4]

# Shallow copy
shallow = copy.copy(original)

# Deep copy
deep = copy.deepcopy(original)

print("Shallow Test:", detect_copy(original, shallow))

# Reset original
original = [1, [2, 3], 4]

print("Deep Test:", detect_copy(original, deep))
```

```python
import copy

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f"Name : {self.name}")
        print(f"Marks: {self.marks}")



student1 = Student("Yash", [80, 90, 95])


student2 = copy.copy(student1)


student3 = copy.deepcopy(student1)


student2.name = "Rahul"
student2.marks.append(100)


student3.name = "Amit"
student3.marks.append(200)

print("Original Student")
student1.display()

print("\nShallow Copy")
student2.display()

print("\nDeep Copy")
student3.display()
```

## 5. Bank Account 
Attributes 
• account_holder  
• balance  
Methods 
• deposit()  
• withdraw()  
Withdraw should happen only if balance is sufficient. 


```python
class Bank_Account:
    def __init__(self,account_holder,balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self):
        amount = int(input("Enter the amount for deposit:"))
        self.balance += amount
        return f"₹{amount} is added and know you balance is {self.balance}"

    def withdraw(self):
        amount = int(input("Enter the amount for withdraw:"))
        if self.balance > amount:
            self.balance-= amount
            return f"₹{amount} is withdraw and know you balance is {self.balance}"
        else:
            print("balance is not safficient for withdrawl")

    def display(self):
        return f"Balance : {self.balance}"
            
    
per = Bank_Account("Yash",50000)

print(per.display())

print(per.deposit())

print(per.withdraw())

print(per.display())
```

## 8. Movie Ticket


```python
class Ticket:
    def __init__(self,movie_name,seats):
        self.movie_name = movie_name
        self.seats = seats
    def book_ticket(self,number):
        if self.seats >= number:
            self.seats -= number
            print(f"Your {number} seats are Booking conform")
        else:
            print(f"{number} Seats are not available. only {self.seats} are available")

    def display(self):
        print(f"movie name {self.movie_name} and sets available is {self.seats}")
        


obj = Ticket("Super 30", 100)

obj.display()

obj.book_ticket(22)

obj.display()


```

## 15. Wallet 


```python
class wallet:
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance

    def purchase(self,amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"purchase allowed enough money")
            print(f"current avilable money is {self.balance}")

        else:
            print(f"Balance is not safficient for purchase balance is {self.balance}")

    def display(self):
        print("owner is :", self.owner)
        print("Balance is :", self.balance)

obj = wallet("Yash", 400)

obj.display()

obj.purchase(500)
            
        
```

## 17. School Fees

```python
class school:
    def __init__(self,student_name,total_fee,paid_fee):
        self.student_name = student_name
        self.total_fee = total_fee
        self.paid_fee = paid_fee

    def pay_fee(self,amount):
        
        if amount + self.paid_fee <= self.total_fee:
            self.paid_fee += amount
            if self.paid_fee == self.total_fee:
                print("Fee paid")
            else:
                print(f"you want to pay remaining {self.total_fee - self.paid_fee}")
                
        else:
            print(f"you pay more {amount+self.paid_fee - self.total_fee} than total fee")


obj = school("Yash",10000,5000)

obj.pay_fee(600)

            

        
            
            
```

## 18. Hospital Patient

```python
class hospital:
    def __init__(self,patient_name,age,temperature):
        self.patient_name = patient_name
        self.age = age
        self.temperature = temperature

    def update_temperature(self,temp):
        if temp >= 102:
            print("Emergency")
        else:
            self.temperature += temp
            print(f"New temperature is :{temp} F")


obj = hospital("Yash",22,100)

obj.update_temperature(99)
obj.update_temperature(122)

print(obj.temperature)
```

## 19. ATM

```python
class ATM:
    def __init__(self,balance):
        self.balance = balance

    def deposit(self):
        amount = int(input("Enter the amount:"))
        if amount > 0:
            
            self.balance+=amount
            print(f"₹{amount} is deposited Balance is ₹{self.balance}")
        elif amount < 0:
            print("Amount is negative")
        else :
            print("Amount is zero")
        

    def withdraw(self):
        amount = int(input("Enter the amount:"))
        if amount > 0:
            if self.balance >=  amount:
                self.balance-= amount
                print(f"₹{amount} is withdrwal and your balance is ₹{self.balance}")
            else:
                print("insufficient balance")
        elif amount < 0:

            print("Amount is negative")
        else:
            print("Amount is zero")


obj = ATM(2000)

obj.deposit()

obj.withdraw()
            
```

## 21. Inventory System 


```python
class Inventory:
    def __init__(self,product_name, stock):
        self.product_name = product_name
        self.stock = stock
        
    def update_stock(self,change):
        if change > 0:
            self.stock += change
            print(f" Add change is stock successfully")
            print(f" Product name :{self.product_name}\n Stock is : {self.stock}")
        elif change < 0:
            self.stock += (change)
            print(f"Remove change in stock successfully")
            print(f" Product name :{self.product_name}\n Stock is : {self.stock}")
        else :
            print(f" Product name :{self.product_name}\n Stock is : {self.stock}")

obj = Inventory("mobile",20)

obj.update_stock(29)
obj.update_stock(-35)

print(obj.stock)
            
```
