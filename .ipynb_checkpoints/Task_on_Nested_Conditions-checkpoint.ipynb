### 1. Take a number as input. If the number is positive, check whether it is even or odd and print the result.

```python
number = int(input("Enter the number:"))

if number > 0:
    if number % 2 == 0:
        print("Even")
    else:
        print("Odd")
else:
    print("Number is zeor or negative")
```

### 2. Take age as input. If age is greater than or equal to 18, check if the person is a
### senior citizen (age >= 60) or adult.

```python
Age = int(input("Enter the age:"))
if Age >= 18:
    if Age >= 60:
        print("Senior citizen")
    else:
        print("Adult")
else:
    print("child")
```

```python
### 3. Take a single character. If it is an alphabet, check whether it is uppercase or lowercase.
char = input("Enter any character:")

if char.isalpha():
    if char.isupper():
        print("Uppercase")
    else:
        print("Lowercase")
else:
    print("please enter alphabet")
        
```

### 5. Take student marks. If marks >= 40 (Pass), then further check if marks >= 75 to print "Pass with Distinction".


```python
marks = int(input("Ente student marks:"))

if marks >= 40:

    if marks >= 75:
        print("pass with Distinction")
    else:
        print("pass")
else:
    print("Faid")
```

### 6.Take a number. If it is divisible by 2, check if it is also divisible by 4.

```python
num = int(input("Enter any number:"))

if num % 2 == 0:
    if num % 4 == 0:
        print("number is divisible by both")
    else:
        print("Number only divisible by 2")

else:
    print("Number is not divisible by both")
    
```
