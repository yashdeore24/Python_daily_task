```python
# 1. Class Name: BankAccount 
# Problem 
# Create a class named BankAccount. 
# Explanation 
# A bank account stores money. The balance should not be directly accessible from outside  the class, so create a private attribute named __balance. 
# Implement the following methods: 
# • deposit(amount)  
# o Deposit money into the account.  
# o Amount must be greater than 0.  
# • withdraw(amount)  
# o Withdraw money.  
# o Withdrawal should happen only if:  
# ▪ amount > 0  
# ▪ amount <= available balance  


class BankAccount():

    def __init__(self,balance):
        self.__balance = balance

    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} is deposited successfully")
        else:
            print(f"{amount} is lessthan the zero ")

    def Withdraw(self,amount):
        if amount > 0:
            if amount <= self.__balance:
                self.__balance -= amount
                print(f"₹{amount} is withdrawl successfully and your balance is ₹{self.__balance}")
            else:
                print("Balance is not sufficiant")
        else:
            print("Please provide the correct or positive  value ")
    
```

# 16.Task

```python
"""16. Class Name: Shape
Problem
Create an abstract class Shape.
Explanation
Declare abstract methods
• calculate_area()

Sumit Sir Notes
• calculate_perimeter()
Create subclasses like
• Circle
• Rectangle
Implement both methods."""

from abc import ABC, abstractmethod
import math

# abstract class
class Shape(ABC):

    @abstractmethod
    def calculate_area(self):
        pass

    @abstractmethod
    def calculate_perimeter(self):
        pass


class Circle(Shape):

    def __init__(self,rad):
        self.__rad = rad

    def calculate_area(self):
        print("Area of circle is : ", math.pi*self.__rad**2)

    def calculate_perimeter(self):
        print("Perimeter of the circule is :",2 * math.pi * self.__rad)
    

class Rectangle(Shape):

    def __init__(self,lenght,width):
        self.__length = lenght
        self.__width = width

    def calculate_area(self):
        print("Area of rectangel is :",self.__length*self.__width)


    def calculate_perimeter(self):
        print("Parimeter of the circule is :",2 * (self.__length + self.__width))


    

c = Circle(5)
c.calculate_area()
c.calculate_perimeter()

print()

r = Rectangle(10, 5)
r.calculate_area()
r.calculate_perimeter()
```

```python
"""22. Class Name: Device
Problem
Create abstract class Device.
Explanation
Abstract methods
• turn_on()
• turn_off()
Implement in
• TV
• Laptop
• Fan
Each device should define its own behavior."""


class Device(ABC):

    # abstract method :
    @abstractmethod
    def turn_on(self):
        pass

    # abstract method :
    @abstractmethod
    def turn_off(self):
        pass

# child class for TV
class TV(Device):

    def turn_on(self):
        print("TV is switched on using the remote. !")

    def turn_off(self):
        print("TV is switched off. !")

# child class for Laptop
class Laptop(Device):

    def turn_on(self):
        print("Laptop is booting up. !")

    def turn_off(self):
        print("Laptop is shutting down !")

# child class of fan 
class Fan(Device):

    def turn_on(self):
        print("Fan starts rotating !")

    def turn_off(self):
        print("Fan stops rotating. !")


```

# 23.Task

```python
"""
23. Class Name: Calculator 
Problem 
Create abstract class Calculator. 
Explanation 
Abstract method 
compute(a, b) 
Implement in 
• AddCalculator  
• MultiplyCalculator 
Sumit Sir Notes 
Each subclass performs a different operation. 

"""
from abc import ABC, abstractmethod

class Calculator(ABC):

    def __init__(self,a,b):
        self.__a = a
        self.__b = b

    @abstractmethod
    def compute(self):
        pass

class AddCalculator(Calculator):
    
    def compute(self):
        print("Addition is",self._Calculator__a + self._Calculator__b)

class MultiplyCalculator(Calculator):

    def compute(self):
        print("Multiplication is",self._Calculator__a * self._Calculator__b)

ADD = AddCalculator(2,4)
MUL = MultiplyCalculator(5,4)
ADD.compute()
MUL.compute()
```

# 25.Task

```python
from abc import ABC, abstractmethod
import math


class Shape3D(ABC):

    @abstractmethod
    def volume(self):
        pass

    @abstractmethod
    def surface_area(self):
        pass


class Cube(Shape3D):

    def __init__(self, side):
        self.side = side

    def volume(self):
        return self.side ** 3

    def surface_area(self):
        return 6 * (self.side ** 2)


class Sphere(Shape3D):

    def __init__(self, radius):
        self.radius = radius

    def volume(self):
        return (4 / 3) * math.pi * (self.radius ** 3)

    def surface_area(self):
        return 4 * math.pi * (self.radius ** 2)


class Cylinder(Shape3D):

    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def volume(self):
        return math.pi * (self.radius ** 2) * self.height

    def surface_area(self):
        return 2 * math.pi * self.radius * (self.radius + self.height)


# Creating objects
cube = Cube(5)
sphere = Sphere(7)
cylinder = Cylinder(3, 8)

# Cube
print("Cube")
print("Volume:", cube.volume())
print("Surface Area:", cube.surface_area())

# Sphere
print("\nSphere")
print("Volume:", sphere.volume())
print("Surface Area:", sphere.surface_area())

# Cylinder
print("\nCylinder")
print("Volume:", cylinder.volume())
print("Surface Area:", cylinder.surface_area())
```
