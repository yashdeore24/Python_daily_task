### 10. Take a year and check if it is a leap year (div by 4 but not 100, or div by 400).



```python
year = int(input("Enter any number:"))

if (year % 4 == 0 and year % 100 == 0) or (year % 400 == 0):
    print(f"{year} is leap")
else:
    print(f"{year} is not leap")
    
```

# Hard Level (15-20)

### Take three sides of a triangle. Check if they form a valid triangle. If yes, also print if it is Equilateral, Isosceles, or Scalene.

```python
s1 = float(input("Enter the fist side:"))
s2 = float(input("Enter the second side:"))
s3 = float(input("Enter the third side:"))


if s1 == s2 and s2 == s3 and s3 == s1:
    print("Equilateral triangle:")
elif s1 == s2 or s2 == s3 or s3 == s1:
    print("Isosceles triangle")
elif s1 != s2 or s2 != s3 or s3 != s1:
    print("Scalene triangle")
elif (s1 + s2) > s3:
    print("Valid triangle")
else:
    print("Not valid")
    
```

### 16. Take student marks in 5 subjects. Calculate percentage and give grade: 90+ → A+, 80-89 → A, 70-79 → B, 60-69 → C, below 60 → Fail.

```python
sub1 = int(input("Enter first subject marks:"))
sub2 = int(input("Enter secound subject marks:"))
sub3 = int(input("Enter third subject marks:"))
sub4 = int(input("Enter fourth subject marks:"))
sub5 = int(input("Enter fifth subject marks:"))

total= sub1+sub2+sub3+sub4+sub5
print("Total marks is:",total)

per = (total / 500) *100
print("percentage is:",per)
if per >= 90:
    print("A+")
elif per >= 80 and per <= 89:
    print("A")
elif per >= 70 and per <= 79:
    print("B")
elif per >=60 and per <=69 :
    print("C")
elif per < 60:
    print("Fail")
```

### 17. Take a number. Check if it is divisible by both 3 and 5. If yes print "FizzBuzz", by 3 print "Fizz", by 5 print "Buzz", else print the number.

### 20. Take three numbers. Print them in ascending order using if-elif-else (without using sort()).

```python
a = int(input("Enter first number:"))
b = int(input("Enter secound number:"))
c = int(input("Enter third number:"))

if a <= b and b <= c:
    if b <= c:
        print(a,b,c)
    else:
        print(a,c,b)
        
elif b <= a and b <= c:
    if a <= c:
        print(b, a, c)
    else: 
        print(b, c,a)
        
elif c <= a  and c <= b:
    if a <= b:
        print(c,a,b)
    else:
        print(c,b,a)

```
