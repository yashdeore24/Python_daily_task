# 2.Task

```python
# 2. Given d = {"a": 1, "b": 2, "c": 3}, print all keys using keys() method.
d = {"a": 1, "b": 2, "c": 3}
print(d.keys())
```

# 3.Task

```python
# Given d = {"a": 1, "b": 2}, print all values using values() method.
d = {"a": 1, "b": 2}
print(d.values())
```

# 4.Task

```python
# 4. Use get() method to safely get the value of key "name" from dictionary d = {"name": "Sumit", "age": 25}.
d = {"name": "Sumit", "age": 25}
print(d.get("name"))
```

# 5.Task

```python
# Create a tuple with 5 numbers and print its type and length.
Tuple = ( 4,5,3,1,2)

print("Type of Tuple variable is:",type(Tuple))
print("length of the Tuple variable is:",len(Tuple))
```


# 6.Task

```python
# Create a tuple tup = (10, 20, 30, 40, 50) and use count() to count how many times 20 appears.
tup = (10, 20, 30, 40, 50)

print(tup.count(20))
```

# 7.Task

```python
# 7. Use index() method on tuple (10, 20, 30, 40) to find the position of 30.
Tuple = (10, 20, 30, 40)

print("The position of 30 in this tuple is:",Tuple.index(30)+1)
```

# 8.Task

```python
# Create an empty dictionary and add 3 key-value pairs using assignment.
dic = {}
print(dic) # empty

dic["Car"] = "Thar"
dic["Modle"] = 2026
dic["color"] = "Black"
print(dic)
```

# 9.Task

```python
# Given d = {"a": 1, "b": 2}, use update() to add {"c": 3, "d": 4}.
d = {"a": 1, "b": 2}

print("Before update:",d)

d.update({"c":3 , "d": 4})

print("Updated:",d)
```
