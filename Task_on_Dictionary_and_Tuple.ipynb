# 23.Task

```python
# Take 4 student names and their marks as input from the user.
# Store them in a dictionary (name as key, marks as value). Then find and
# print the student with the highest marks.
student1 = input("Enter name:")
marks1 = int(input("Enter the marks:"))
student2 = input("Enter name:")
marks2 = int(input("Enter the marks:"))
student3 = input("Enter name:")
marks3 = int(input("Enter the marks:"))
student4 = input("Enter name:")
marks4 = int(input("Enter the marks:"))

dic = {}

dic.update({student1:marks1 , student2:marks2, student3:marks3, student4:marks4})

highest_student = max(dic,key = dic.get)

print("Student with highest marks:")
print(highest_student, ":", dic[highest_student])
```

```python
"""(Hard) Create a dictionary info = {"name": "Aman", "age": 21, "city":
"Mumbai"}. Make a copy of it. Clear the original dictionary. Then add a
new key "country": "India" in the copy and print both dictionaries."""

info = {"name": "Aman", "age": 21, "city":"Mumbai"}

Copy = info.copy()

info.clear()

Copy.update({"country": "India"})

print("Original:",info)
print("Copy:",Copy)
```

# 25.Task

```python
"""(Hard) Given data = (10, 20, 30, 20, 40, 20, 50), convert it into a
dictionary where key is the number and value is its count of occurrences
(using tuple methods + dictionary)"""
data = (10, 20, 30, 20, 40, 20, 50)

result = {}

result[10] = data.count(10)
result[20] = data.count(20)
result[30] = data.count(30)
result[40] = data.count(40)
result[50] = data.count(50)

print(result)
```
