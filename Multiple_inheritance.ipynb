## 1. Create two parent classes Father and Mother. Create a child class Child that inherits from both.

```python
class Father:

    # constructor
    def __init__(self,surname,**key):
        super().__init__(**key)
        self.surname = surname
        
class Mother:

    # constructor
    def __init__(self,skill, **key):
        super().__init__(**key)
        self.skill = skill
        


class Child(Father,Mother):

    # constructor
    def __init__(self,name,surname,skill):
        super().__init__(surname = surname ,skill = skill)
        self.name = name

    def display(self):
        print("Name : ",self.name)
        print("Surname :",self.surname)
        print("skill :",self.skill)

obj = Child("Yash","Deore","Cooking")

obj.display()
```

```python
class Father:

    def __init__(self, father_surname, **kwargs):
        super().__init__(**kwargs)
        self.surname = father_surname


class Mother:

    def __init__(self, mother_skill, **kwargs):
        super().__init__(**kwargs)
        self.skill = mother_skill


class Child(Father, Mother):

    def __init__(self, child_name, child_surname, child_skill):
        super().__init__(
            father_surname=child_surname,
            mother_skill=child_skill
        )
        self.name = child_name

    def display(self):
        print("Name :", self.name)
        print("Surname :", self.surname)
        print("Skill :", self.skill)


obj = Child("Yash", "Deore", "Cooking")
obj.display()
```

## 11. Create two parent classes with static methods. Create a child class that inherits from  both and calls the static methods. 


```python
class Parent:

    @staticmethod 
    def display(self):
        return f"It's a parent class static method"

class GrandParent:

     @staticmethod 
     def Display(self):
         return f"It's a grandparent class static method"

class Child(Parent,GrandParent):

    def show(self):
        print(Parent.display(self))
        print(GrandParent.Display(self))
        
    

    
obj = Child()

obj.show()
```

## 17. Create class Calculator with static method and another class Display. Create child class that uses both.

```python
class Calculator:

    @staticmethod
    def mul(a, b):
        return f"Multiplication of {a} and {b} is {a * b}"


class Display:

    def display(self, result):
        print(result)


class Child(Calculator, Display):

    def __init__(self, a, b):
        self.a = a
        self.b = b

    def pne(self):
        result = self.mul(self.a, self.b)   # inherited static method
        self.display(result)                # inherited instance method


obj = Child(3, 5)
obj.pne()
```

## 19. Create two abstract classes Shape and Color. Create a concrete class that inherits from both and implements abstract methods.

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def shape(self):
        pass
    

class Color(ABC):

    @abstractmethod
    def color(self):
        pass

class Concrete(Shape,Color):


    def shape(self):
        print("Shape is cube")

    def color(self):
        print("Color is red")

obj=Concrete()

obj.shape()
obj.color()
    


```

## 22. Define abstract class Readable and abstract class Writable. Create class File that implements both.

```python
class Readable(ABC):

    @abstractmethod
    def read(self):
        pass

class Writable(ABC):

    @abstractmethod
    def write(self):
        pass

class File(Readable,Writable):

    def read(self):
        print("Read the notes")

    def write(self):
        print("write the notes")
        
obj = File()
obj.read()
obj.write()
```
