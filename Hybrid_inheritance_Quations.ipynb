```python
"""
1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits  from both B and C). 
Concept 
Hybrid Inheritance (Combination of Single + Multiple Inheritance) 
What to do 
• Create parent class A.  
• Create class B inheriting from A.  
• Create another class C.  
• Create class D inheriting from both B and C.  
How to solve 
1. Create class A.  
2. Add one method.  
3. Create class B(A).  
4. Create class C.  
5. Create class D(B, C).  
6. Create an object of D.  
7. Call methods from all parent classes.  


"""

class A:
    def show_A(self):
        print("It is A class")
        
class B(A):
    def show_B(self):
        print("It is a B class")
        
class C:
    def show_C(self):
        print("It is a C class")
        
class D(B,C):
    def show_D(self):
        print("It is a D class")
        


obj = D()
obj.show_A()
obj.show_B()
obj.show_C()
obj.show_D()
```

```python
class Person:
    def __init__(self, name, **kwargs):
        super().__init__(**kwargs)
        self.name = name

class Employee(Person):
    def __init__(self, salary, **kwargs):
        super().__init__(**kwargs)
        self.salary = salary

class Student(Person):
    def __init__(self, marks, **kwargs):
        super().__init__(**kwargs)
        self.marks = marks

class TeachingAssistant(Employee, Student):
    def __init__(self, name, salary, marks, department):
        super().__init__(name=name, salary=salary, marks=marks)
        self.department = department

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Marks:", self.marks)
        print("Department:", self.department)

obj = TeachingAssistant("Yash", 40000, 85, "DS")
obj.display()
```

```python
"""
4. Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO). Concept 
Method Resolution Order (MRO) 
What to do 
Create 
A 
| 
B 
C 
D(B,C) 
Print 
D.mro() 
How to solve 
1. Create four classes.  
2. Give same method name in multiple classes.  
3. Print ClassName.mro().  
4. Observe Python's searching order.

"""

class A:
    def show(self):
        print("class A")
        
class B(A):
    def show(self):
        print("class B")
class C:
    def show(self):
        print("class C")
class D(B,C):
    def show(self):
        print("class D")

obj=D()
obj.show()
print(A.mro())
print(B.mro())
print(C.mro())
print(D.mro())



```

```python
"""
5. Create class A, B, C, D and solve the Diamond Problem using super(). Concept
Sumit Sir Notes 
Diamond Problem 
Hierarchy 
 A 
 / \ 
 B C 
 \ / 
 D 
What to do 
Use super() everywhere. 
How to solve 
1. Create constructor in A.  
2. Call super() in B.  
3. Call super() in C.  
4. Call super() in D.  
5. Create object of D.  
6. Observe A constructor runs only once.  

"""
class A:
    def __init__(self):
        super().__init__()
        print("It is a class A")
        
class B(A):
    def __init__(self):
        super().__init__()
        print("It is a Class B")
        
class C(A):
    def __init__(self):
        super().__init__()
        print("It is a class C")
        
class D(B,C):
    def __init__(self):
        super().__init__()
        print("It is a class D")

obj = D()
```

```python
"""
6. Create a banking system using hybrid inheritance. Concept 
Hybrid + Real-world Example 
What to do 
Account 
 | 
Savings 
Loan 
SavingsLoanAccount
Sumit Sir Notes 
How to solve 
1. Account stores account number.  
2. Savings stores balance.  
3. Loan stores loan amount.  
4. SavingsLoanAccount inherits both.  5. Print all details.  

"""

class Account:
    def __init__(self,account_number,**keyargs):
        super().__init__(**keyargs)
        self.Account_number = account_number
        
class Savings (Account):
    
    def __init__(self,balance,**keyargs):
        super().__init__(**keyargs)
        
        self.Balance = balance
        
class Loan:
    def __init__(self,loan_amount,**keyargs):
        super().__init__(**keyargs)
        self.Loan_amount = loan_amount
        
class SavingsLoanAccount(Savings,Loan):

    def __init__(self,account_number,balance,loan_amount):
        super().__init__(
            account_number = account_number,
            balance = balance,
            loan_amount = loan_amount)
        

    def display(self):
        print("Account number:",self.Account_number)
        print("Balance is :",self.Balance)
        print("Loan Amount:",self.Loan_amount)

obj = SavingsLoanAccount(234242,2000000,100000)
obj.display()

    
```

```python
"""
19. Create a hybrid inheritance structure and apply Duck Typing. Concept 
Hybrid + Duck Typing
Sumit Sir Notes 
What to do 
Create hybrid inheritance. 
Create another unrelated class with same method. 
How to solve 
1. Create hybrid hierarchy.  
2. Final child has method show().  
3. Another unrelated class also has show().  
4. One function calls show() for both.

"""
class A:
    def show(self):
        print("It is a A class")
class B(A):
     def show(self):
        print("It is a B class")
class C:
     def show(self):
        print("It is a C class")
class D(B,C):
     def show(self):
        print("It is a D class")

class Unrelated:
     def show(self):
        print("It is a Unrelated class")

obj = D()
obj1 = Unrelated()

def Show(OBJ):
    OBJ.show()

Show(obj)
Show(obj1)

    
```
