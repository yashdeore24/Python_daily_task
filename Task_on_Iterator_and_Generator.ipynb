## Problem 25 :Create a custom iterator that prints the Fibonacci sequence up to 10  terms. 

```python
class Fibonacci:
    def __init__(self, terms):
        self.terms = terms
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.terms:
            value = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return f"{value}"
        else:
            raise StopIteration


fib = Fibonacci(10)

for num in fib:
    print(num)
```

```python
## Problem 27 : Create a generator that yields all prime numbers between 1 and 100. 

def number():

    for i in range(2,100):
        prime = True
        
        for j in range(2,i):
            if i % j == 0:
                prime = False
                break
        
        if prime:
            yield i
        
                
            
      

generator = number()

for i in generator:
    print(i,end=",")
```
