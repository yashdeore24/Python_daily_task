## Q15 Suppose the Balance column has missing values. Replace them using the mean of the Balance column. 


```python
df['Bank_Balance'] = df['Bank_Balance'].fillna(df['Bank_Balance'].mean())
print(df['Bank_Balance'].head())
```

```python
df['Bank_Age'] = df['Bank_Age'].fillna(df['Bank_Age'].median())
print(df['Bank_Age'].head())
```
