## 30.Explain the difference in readability and performance between a normal for  loop and list comprehension. When should you avoid list comprehensions?

```python
numbers = [1, 2, 3, 4, 5]

result = [i * 2 for i in numbers]

print(result)
```

```python
numbers = [1, 2, 3, 4, 5]

result = []

for i in numbers:
    result.append(i * 2)

print(result)
```
