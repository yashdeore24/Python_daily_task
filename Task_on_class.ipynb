```python
class BankAccount:

    history = []

    def __init__(self, balance):
        self.__balance = balance

    # Getter
    def get_balance(self):
        return self.__balance

    # Setter
    def set_balance(self, new_balance):

        if new_balance < 1000:
            print("Balance cannot be less than 1000.")
            return

        if new_balance > 100000:
            bonus = (5 / 100) * new_balance
            new_balance += bonus
            print("5% bonus added.")

        self.__balance = new_balance

        BankAccount.history.append(self.__balance)

        print("Balance updated successfully.")
```

## 3.Task

```python
class Student:

    total_students = 0

    def __init__(self, marks):
        self.__marks = marks
        self.status = "Passed"
        Student.total_students += 1

    # Getter
    def get_marks(self):
        return self.__marks

    # Setter
    def set_marks(self, new_marks):

        # Check exactly 5 subjects
        if len(new_marks) != 5:
            print("Marks list must contain 5 subjects.")
            return

        # Check marks between 0 and 100
        for mark in new_marks:
            if mark < 0 or mark > 100:
                print("Marks must be between 0 and 100.")
                return

        # Previous average
        previous_avg = sum(self.__marks) / 5

        # New average
        new_avg = sum(new_marks) / 5

        # Average cannot decrease
        if new_avg < previous_avg:
            print("Average cannot decrease.")
            return

        # Update marks
        self.__marks = new_marks

        # Check pass/fail
        if min(new_marks) < 40:
            self.status = "Failed"
        else:
            self.status = "Passed"

        print("Marks updated successfully.")
```
