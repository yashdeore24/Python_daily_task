multiplies any number by n. (Use closure concept). Then create double = make_multiplier(2)

### 3.(Hard) Write a function smart_divide(func) as a decorator that:
• Takes a function as argument
• Checks if the second argument is zero before division
• If zero, prints "Cannot divide by zero!" and returns None
• Otherwise, calls the original function

```python
def smart_divide(func):
    def wrapper(a,b):
        if b == 0:
            print("Cannot divide by zero!")
            return None
        return func(a,b)
    return wrapper

@smart_divide
def devide(a,b):
   return a/ b

devide(4,2)

```

### 4. (Hard) Create a function flexible_calc(*args, operation="add", **kwargs) that:
• Can take any number of numbers
• Supports operations: add, subtract, multiply, divide
• If round_result=True in kwargs, round the final answer to 2 decimal places
• Handle all edge cases

```python
def flexible_calc(*args, operation="add", **kwargs):

    # Edge case: No numbers provided
    if len(args) == 0:
        return "No numbers provided!"

    # Addition
    if operation == "add":
        result = 0
        for i in args:
            result += i

    # Subtraction
    elif operation == "subtract":
        result = args[0]
        for i in args[1:]:
            result -= i

    # Multiplication
    elif operation == "multiply":
        result = 1
        for i in args:
            result *= i

    # Division
    elif operation == "divide":
        result = args[0]
        for i in args[1:]:
            if i == 0:
                return "Cannot divide by zero!"
            result /= i

    # Invalid operation
    else:
        return "Invalid operation!"

    # Round result if requested
    if kwargs.get("round_result", False):
        result = round(result, 2)

    return result


# Test Cases
print(flexible_calc(2, 3, 4))                               # Add (default)
print(flexible_calc(10, 3, 2, operation="subtract"))
print(flexible_calc(2, 3, 4, operation="multiply"))
print(flexible_calc(20, 5, operation="divide"))
print(flexible_calc(20, 3, operation="divide", round_result=True))
print(flexible_calc(20, 0, operation="divide"))
print(flexible_calc(operation="add"))
print(flexible_calc(2, 3, operation="square"))
```

### 5. (Hard) Write a recursive function flatten_list(nested_list) that takes a nested list (list inside
list) and returns a single flat list. Example: flatten_list([1, [2, 3], [4, [5, 6]]]) → [1, 2, 3, 4, 5, 6]

```python
def flatten_list(nested_list):
    result = []

    for item in nested_list:
        if isinstance(item, list):
            result.extend(flatten_list(item))   # Recursive call
        else:
            result.append(item)

    return result


# Test
print(flatten_list([1, [2, 3], [4, [5, 6]]]))
print(flatten_list([[1, 2], [3, [4, 5]], 6]))
print(flatten_list([[[1]], 2, [[3, 4], 5]]))
```
