## 1. Create a simple line plot using:

```python
import matplotlib.pyplot as plt
X = [1,2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)
```

## 2. Create a line plot and add an appropriate title.

```python
X = [1,2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)
plt.title("Line Plot")
plt.show()
```

## 3. Create a line plot and label both the X-axis and Y-axis.

```python
X = [1,2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)
plt.title("Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

## 4. Plot two lines on the same graph and add a legend to identify them.

```python
X = [1,2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)
plt.title("Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend(["Line"])
plt.show()
```

## 5. Change the line color, line style, and marker style of a line plot.

```python
X = [1,2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y,
        color = "green",
        linestyle = "-",
        marker = "v")
plt.title("Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend(["Line"])
plt.show()
```

## Create a line plot using NumPy values generated from np.arange().
