## 12. Read a file and print only the lines that contain the word “Python”.

```python
with open("Y.txt","w") as file:
    file.write("Python is a programming language\nit is a high level language\npython develop in 1992")
with open("Y.txt","r") as file:
    for line in file:
        words = line.split()
        for i in words:
            if i.lower() == "python":
                print(line)
        
```

## 14. Count how many times a particular word appears in a file. 

```python
count =0
with open("Y.txt","r") as file:
    for line in file:
 
        words = line.split()
        for i in words:
            if i.lower() == "python":
                count+=1
print(count)
```

## 15. Read a file and display the longest word in it. 


```python
largest = "i"

with open("Y.txt","r") as file:
    for line in file:
 
        words = line.split()
        for i in words:
            if len(largest) < len(i):
               largest = i
print(largest)
```

## 17. Store a list of numbers in a file, then read and calculate their sum.

```python


Sum =0
with open("File.txt","w+") as f1:
    f1.write("1,2,3,4,5,6")
    f1.seek(0)
    print(f1.read())

with open("File.txt","r") as f2:
    for line in f2:
        words = line.split(",")
       
        for i in words:
            Sum += int(i)

print(Sum)

    
```

## 21. Create a simple phonebook. Take name and phone number from user and store in a file.



```python
with open("phonebook.txt","w") as f1:
    f1.write("Name    ,Phone\n")
    f1.write("Yash    ,8484975320\n")
    f1.write("Prashant,9322465171")

with open("phonebook.txt","r") as f1:
    for line in f1:
        name,phone = line.strip().split(",")
        print(f"{name}\t{phone:}")
            
        
        
```
