## 22. Write a program that reads a log file and counts the number of “ERROR” messages

```python

with open("app.log", "w") as file:
    file.write("INFO Program started\n")
    file.write("ERROR File not found\n")
    file.write("WARNING Low memory\n")
    file.write("ERROR Database connection failed\n")


count = 0

with open("app.log", "r") as file:
    for line in file:
        if "ERROR" in line:
            count += 1

print("Number of ERROR messages:", count)
```

## 23. Take multiple lines of text from user until they type “STOP”, then save all lines in a file. 


```python
with open("Text.txt","w") as file:
    print("Loop stop when you give input 'stop'")
    
    while True:
        data = input("Enter the input:")
        if data.lower()  == "stop":
            break
        file.write(data)
        file.write("\n")


with open("Text.txt","r") as file:
    print(file.read())
    
```

## 24. Write a program to encrypt the content of a file (simple character shifting) and save it in  another file. 

```python
with open("Text.txt","r") as file:
    data = file.read()

encripted = ""

for ch in data:
    encripted += chr(ord(ch) + 1)

with open("encrypted.txt","w") as file:
    file.write(encripted)

print("File encrypted successfully!")
```

## 25. Read a file containing student marks, calculate average, and write the result in another  file. 


```python
with open("marks.txt", "w") as file:
    file.write("45\n78\n90\n67\n80")

total = 0
count = 0

with open("marks.txt", "r") as file:
    for line in file:
        total += int(line.strip())
        count += 1

average = total / count

with open("result.txt", "w") as file:
    file.write(f"Average Marks = {average}")

with open("result.txt","r") as file:
    print(file.read())
```

## 26. Create a program that keeps a record of daily expenses in a file and can show total  expenses.

```python
with open("expenses.txt", "a") as file:
    while True:
        amount = input("Enter expense (or type STOP to finish): ")

        if amount.lower() == "stop":
            break

        file.write(amount + "\n")

total = 0

with open("expenses.txt", "r") as file:
    for line in file:
        total += float(line.strip())

print("Total Expenses =", total)
```

## 27. Write a program to remove all blank lines from a file. 


```python
with open("input.txt", "w") as file:
    file.write("Yash \n\n")
    file.write("Bhalchandra\n\n")
    file.write("Deore\n\n")
    file.write("Yash \n\n")


with open("input.txt", "r") as file:
    lines = file.readlines()


with open("output.txt", "w") as file:
    for line in lines:
        if line.strip() != "":
            file.write(line)

with open("output.txt", "r") as file:
    print(file.read())
```

## 28. Read a file and write only the unique lines into another file (remove duplicates)

```python

seen = set()

with open("input.txt", "r") as infile, open("output.txt", "w") as outfile:
    for line in infile:
        if line not in seen:
            outfile.write(line)
            seen.add(line)

with open("output.txt", "r") as outfile:
    print(outfile.read())

print("Duplicate lines removed successfully.")
```

## 29. Create a menu-driven program: 
• Write to file 
• Read file 
• Append to file 
• Exit 


```python
while True:
    print("\n--- File Menu ---")
    print("1. Write to file")
    print("2. Read file")
    print("3. Append to file")
    print("4. Exit")

    choice = input("Enter your choice (1-4): ")

    if choice == "1":
        data = input("Enter text to write: ")
        with open("myfile.txt", "w") as file:
            file.write(data + "\n")
        print("Data written successfully.")

    elif choice == "2":
        try:
            with open("myfile.txt", "r") as file:
                print("\\nFile Content:")
                print(file.read())
        except FileNotFoundError:
            print("File does not exist.")

    elif choice == "3":
        data = input("Enter text to append: ")
        with open("myfile.txt", "a") as file:
            file.write(data + "\n")
        print("Data appended successfully.")

    elif choice == "4":
        print("Exiting program...")
        break

    else:
        print("Invalid choice! Please enter 1-4.")
```

## 30. Write a program that handles file exceptions (FileNotFoundError, PermissionError)  properly while reading and writing files

```python
filename = input("Enter file name: ")

try:
    with open(filename, "r") as file:
        content = file.read()
        print("\nFile Content:")
        print(content)

except FileNotFoundError:
    print("Error: File not found.")

except PermissionError:
    print("Error: You do not have permission to read this file.")

except Exception as e:
    print("Unexpected error:", e)


try:
    with open(filename, "a") as file:
        text = input("\nEnter text to append: ")
        file.write(text + "\n")
        print("Data written successfully.")

except PermissionError:
    print("Error: You do not have permission to write to this file.")

except Exception as e:
    print("Unexpected error:", e)
```
