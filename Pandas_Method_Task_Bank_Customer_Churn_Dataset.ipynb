```python
Sum = 0
count = 0

while True:
    num = int(input())
    if num < 0:
        break
    Sum +=num
    count+=1

print("Sum:",Sum)
print("Average",Sum/count)
```

## 11. Create class BankAccount with private __balance. Add deposit and withdraw  methods. Raise custom exception if balance becomes negative

```python
class NegativeNumberError(Exception):
    pass
    
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance

    def deposit(self):
        amount  = int(input("Enter amount for deposit:"))
        if amount > 0:
            self.__balance += amount
            print(f"{amount} is added successfully and your balance is {self.__balance}")
            
        else:
            raise NegativeNumberError("You enter the negative number ")

    def withdraw(self):
        amount  = int(input("Enter amount for withdraw:"))
        if amount > 0:
            if self.__balance > amount :
                self.__balance -= amount
                print(f"{amount} is withdraw successfully and your balance is {self.__balance}")
            
            else:
                print("insafficient balance")
        else:
            raise NegativeNumberError("You enter the negative number ")


try :

    obj = BankAccount(12000)
    obj.deposit()
    obj.withdraw()

except NegativeNumberError as e:
    print(e)

            

    

        
        
```

## 14. Create class Rectangle with private length and width. Add method to calculate  area and perimeter

```python
class Rectangle:
    def __init__(self,length,width):
        self.__length = length 
        self.__width = width

    def area(self):
        print("Area of Rectangel:",self.__length * self.__width)

    def perimeter(self):
        print("Parameter of the rectange:",2*(self.__length + self.__width))


obj = Rectangle(2,4)
obj.area()
```

## 15. Create single level inheritance: Person → Student. Use super() in constructor. 

```python
class Person:
    def __init__(self,name):
        self.name = name

    def show(self):
        return f"Name : {self.name}"
        
class Student(Person):
    def __init__(self,name,age):
        super().__init__(name)
        self.age = age

    def display(self):
        print(super().show())
        print("age :",self.age)


obj =Student("Yash",22)
obj.display()
```
