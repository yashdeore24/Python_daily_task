# 6.Task

```python
digit = 0
rev = 0
num = int(input("Enter the number:"))
original = num

while num > 0:
    digit = num % 10
    rev = digit + rev * 10
    num = num // 10

if original == rev:
    print("Palindrome")
else:
    print("not Palindrome")
```

# 7.task

```python
num = int(input("Enter any number: "))

ori = num
temp = num
count = 0
armstrong = 0

# Count digits
while temp > 0:
    temp = temp // 10
    count += 1

# Calculate Armstrong sum
while ori > 0:
    digit = ori % 10
    armstrong += digit ** count
    ori = ori // 10

print("Armstrong Sum =", armstrong)

# Check Armstrong
if armstrong == num:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```
