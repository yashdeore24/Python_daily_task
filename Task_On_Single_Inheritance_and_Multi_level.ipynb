## 1.Task

```python
"""
1. Class Name: Person → Student
Problem
Create a parent class named Person and a child class named Student.
Explanation
The Person class should store:
• name
• age
The Student class should inherit from Person and add:
• marks
Create a method in Student to display the student's name, age, and marks.
"""

class Person:

    def __init__(self,name,age):
        self.__Name = name
        self.__Age = age

    def person_info(self):
        return f"Person Name : {self.__Name}  \nPerson Age : {self.__Age}"



class Student(Person):

    def __init__(self,name,age,marks):
        super().__init__(name,age)
        self.marks = marks

    def display(self):
        print("marks :", self.marks)
        print(self.person_info())


std = Student("Yash",22,52)

std.display()
```

## 2.Task

```python
"""
Class Name: Vehicle → Car
Problem
Create a parent class named Vehicle.
Explanation
The Vehicle class should contain a method:
• start()
Create a child class Car that:
• Adds a brand attribute.
• Overrides the start() method to display a car-specific starting message.
"""


class Vehicle:


    def start(self):
        print("Vehicle is starting...")



class Car(Vehicle):

    
    def __init__(self, brand):
        self.brand = brand

    def start(self):
        print(f"{self.brand} car starts with a key or push button.")



car1 = Car("")


car1.start()
```

```python
"""
5. Class Name: Shape → Circle Problem 
Create a parent class named Shape. Explanation 
Create a method: 
• area()  
Create child class: 
• Circle 
Sumit Sir Notes 
Store: 
• radius  
Override area() to calculate the area of a circle. 


"""
import math
class Shape:


    def area(self):
        print("area is ...")



class Circle(Shape):

    
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        print(f"area of circle is {math.pi*self.radius**2}")



c = Circle(2)


c.area()
```
