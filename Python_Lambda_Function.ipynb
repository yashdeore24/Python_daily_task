## 11.Use reduce() with lambda to find the sum of all numbers.

```python
from functools import reduce

numbers = [10,20,30,40]

result = reduce(lambda x, y: x + y, numbers)

print(result)
```

## 12.Use reduce() with lambda to find the product of all numbers.

```python
from functools import reduce

numbers = [2,3,4,5]

result = reduce(lambda x, y: x * y, numbers)

print(result)
```

## 30.Task

```python




from functools import reduce

students = [
    {"name":"Ram","marks":[70,80,90]},
    {"name":"Amit","marks":[90,92,94]},
    {"name":"Neha","marks":[60,65,70]},
    {"name":"Raj","marks":[88,84,90]}
]

# 1. Calculate each student's average marks
averages = list(map(
    lambda s: {
        "name": s["name"],
        "average": sum(s["marks"]) / len(s["marks"])
    },
    students
))

print("Average Marks:")
print(averages)

# 2. Sort students by average marks (highest first)
sorted_students = sorted(
    averages,
    key=lambda x: x["average"],
    reverse=True
)

print("\nSorted Students:")
print(sorted_students)

# 3. Find the topper
topper = max(
    averages,
    key=lambda x: x["average"]
)

print("\nTopper:")
print(topper)

# 4. Filter students whose average is greater than 80
filtered_students = list(filter(
    lambda x: x["average"] > 80,
    averages
))

print("\nAverage > 80:")
print(filtered_students)

# 5. Print only the names of the filtered students
names = list(map(
    lambda x: x["name"],
    filtered_students
))

print("\nNames:")
print(names)
```
