```python
# Python
# Take a word from user.
# Check if the word is a palindrome or not.
# (Example: "madam", "radar")

word = input("Enter any word")

str_reverse = word[::-1]
if word == str_reverse:
    print("Word is a Palindrome ")
else:
    print("Word is not Palindrome")
```

```python
# Python
# Take your city name as input.
# Print the following:
# 1. City name in reverse order
# 2. Number of vowels in the city name (a,e,i,o,u)
# 3. City name without first and last character
text = input("Enter the city name")

print("city name in reverse order:",text[::-1])


count = (
    text.lower().count("a") +
    text.lower().count("e") +
    text.lower().count("i") +
    text.lower().count("o") +
    text.lower().count("u")
)

print("Number of vowels:", count)

print("city name without first and last character:", text[1:-1])

```

```python
# 5. Mixed (String + Logical)
# Python
# Take a string from user.
# Check if:
# - The string length is greater than 5

# Sumit Sir Notes
# - AND it starts with 'A' or 'a'
# Print "Valid String" or "Invalid String"

user_str = input("Enter any String:")

length = len(user_str)

if length > 5 and ( user_str[0] == 'A' or user_str[0] == 'a' ):
    print("Valid String")
else:
    print("Invalid String")
```
