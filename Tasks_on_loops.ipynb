# 2. (Hard) Take a number n from the user. Using while loop, print all prime numbers between 2 and n.

```python
n = int(input("Enter a number: "))

num = 2

while num <= n:
    is_prime = True
    i = 2

    while i < num:
        if num % i == 0:
            is_prime = False
            break
        i += 1

    if is_prime:
        print(num)

    num += 1
```

# 5. (Hard) Create a number guessing game using while loop:
• Generate a random number between 1 to 100.
• Give the user 7 chances to guess.
• After each guess, tell whether the number is "Too High" or "Too Low".
• If user guesses correctly, print "Congratulations!" and stop.
• If chances are over, reveal the number.

```python
import random
secret = random.randint(1,100)
chance = 1

while chance <= 7:
    guess = int(input("Guess the number:"))

    if guess == secret:
        print("congratulations")
        break

    elif guess < secret :
        print("Too Low")

    else:
        print("Too High")

    chance +=1
if chance == 7:
    print("number is:",secret)
```
