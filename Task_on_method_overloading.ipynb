## Method Overloading Questions

```python
"""
1. Create a Calculator class that can add 2 numbers or 3 numbers using the same method  name. 
Concept 
Method Overloading (Python uses default arguments or *args to achieve overloading.) What to do 
• Create a class Calculator.  
• Create one method add().  
• The method should work for either 2 numbers or 3 numbers.  
How to solve 
1. Create class Calculator.  
2. Define method add().  
3. Use default arguments or *args.  
4. Check how many numbers are provided.  
5. Return the sum.  

"""

class Calculator:

    def add(self,a = 0 ,b = 0 , c = 0 ,d = 0):
        return a + b + c + d

obj = Calculator()
obj.add(2,3)
```

```python
"""
2. Create an Area class that calculates the area of a square and rectangle using the same  method. 
Concept 
Method Overloading using default arguments. 
What to do 
• One parameter → Square.  
• Two parameters → Rectangle.  
How to solve 
1. Create class Area.  
2. Define method calculate().  
3. If only one value is provided, calculate the square's area.  
4. If two values are provided, calculate the rectangle's area.  
5. Return the result. 

"""

class shape:

    def area(self, Str,a = 1 ,b = 1 ):
        if b == 1:
            print(f"Area of {Str} : {a * a}")
        else :
            print(f"Area of {Str} : {a * b}")
            
        

obj = shape()
obj.area("Square",2)
obj.area("Rectangle",2,4)
```

```python
"""
3. Create a Student class with a details() method that accepts different numbers of  arguments. 
Concept 
Method Overloading 
What to do 
The method should work with: 
• Name  
• Name + Age  
• Name + Age + City  
How to solve 
1. Create class Student.  
2. Create method details().  
3. Use default arguments.  
4. Print the information based on the provided values

"""

class Student:

    def details(self,Name = "Not deffind",Age = "Not deffind", City = "Not deffind"):
        
        print(f"Name is : {Name}\nAge is : {Age}\nCity is : {City}")

obj = Student()
obj.details("Yash")
print()
obj.details("Yash",22)
print()
obj.details("Yash",22,"Dondaicha")
```

```python
"""
4. Create a MathOperations class that multiplies any number of values. Concept 
Method Overloading using *args 
What to do 
Multiply: 
• 2 numbers  
• 3 numbers  
• 4 numbers  
• More numbers  
How to solve 
1. Create class MathOperations.  
2. Define method multiply(*args

"""
class Multiplies:

    def multiplay(self,*args):
        result = 1
        for i in args:
            result *= i
        return f"Multiplication is :{result}"

obj = Multiplies()
print(obj.multiplay(1,3,2,4,5))
```

```python
"""
5. Create a Display class that prints different messages based on the argument type. Concept 
Method Overloading + Type Checking 
What to do 
Accept: 
• Integer  
• String  
• List  
How to solve 
1. Create class Display.  
2. Create method show().  
3. Use isinstance().  
4. Print different messages depending on the object's type.  

"""

class Display:

    def show(self,argument):
        if isinstance(argument , int):
            print(f" {argument}  is Integer ")
            
        elif isinstance(argument , str):
            print(f"{argument} is String")
            
        elif  isinstance(argument,list):
            print(f"{argument} is list")

        elif isinstance(argument,dict):
            print(f"{argument} is Dictionary")

        elif isinstance(argument , float):
            print(f"{argument} is float")

        else :
            
            print("Type not detact")

obj = Display()
obj.show(2.2)
obj.show(2)
obj.show("Yash")
obj.show({"Name" : "Yash"})
obj.show([1,2,3,4])
            
```

```python
"""6. Create a Volume class that calculates the volume of a cube and cuboid. Concept 
Method Overloading 
What to do 
• One parameter → Cube  
• Three parameters → Cuboid  
How to solve 
1. Create class Volume.  
2. Create method calculate(). 
Sumit Sir Notes 
3. If one value is given, calculate the cube's volume.  4. Otherwise, calculate the cuboid's volume. """

class Volume:
    def calculate(self,*args):
        
            if len(args) == 1:
                print(f"Volume of Cube is : {args[0] ** 3}")
            
            elif len(args) == 3:
                print(f"Volume of Cuboid is : {args[0] * args[1] * args[2]}")

            else:
                print("Please enter the valid arguments")

obj = Volume()
obj.calculate(2)
obj.calculate(1,2,3)
```

```python
"""
7. Create a PrintData class that prints one or many values. Concept 
Method Overloading using *args 
What to do 
Print: 
• One value  
• Multiple values  
How to solve 
1. Create class PrintData.  
2. Create method print_data(*args).  
3. Loop through the arguments.  
4. Print each value.  


"""
class PrintData:

    def print_data(self,*args):
        for i in args:
            print(i)
            
obj= PrintData()
obj.print_data(1,2,3,"Yash")
```

```python
"""
8. Create an Animal class with sound(). Override it in Dog and Cat. Concept 
Method Overriding 
What to do 
• Animal → Generic sound  
• Dog → Bark  
• Cat → Meow  
How to solve 
1. Create parent class Animal.  
2. Define method sound().  
3. Inherit Dog and Cat. 
Sumit Sir Notes 
4. Override sound().  
5. Create objects and call the method.  

"""
class Animal:

    def sound(self):
        print("noice")

class Dog(Animal):

    def sound(self):
        print("Bark")

class Cat(Animal):

    def sound(self):
        print("Meow")

obj = Cat()
obj1 = Dog()
obj.sound()
obj1.sound()
```

```python
"""10. Create an Employee class and override calculate_salary() in Manager. Concept 
Method Overriding 
What to do 
• Employee → Normal salary  
• Manager → Salary + Bonus  
How to solve 
1. Create parent class.  
2. Define calculate_salary().  
3. Inherit Manager.  
4. Override the method.  
5. Print the manager's salary
"""

class Employee:

    def __init__(self,salary):
        self.salary = salary

    def calculate_salary(self):
        print(f"salary of Employee {self.salary}")
        
class Manager(Employee):

    def calculate_salary(self):
        print(f"original salary  is {self.salary} and with 20% bonus salary is {self.salary + ( self.salary * 0.2)}")

obj = Manager(40000)
obj.calculate_salary()
```

## Operator Overloading Questions (15-20)

```python
"""
5. Create a Point class and overload the + operator. Concept 
Operator Overloading (__add__) 
What to do 
Add two points together. 
How to solve 
1. Create class Point. 
Sumit Sir Notes 
2. Store x and y coordinates.  
3. Overload __add__().  
4. Return a new point with added coordinates.  

"""
class Point:
     def __init__(self,grade):
        self.Grade = grade
        
 
     def __add__(self,other):
        result  = self.Grade + other.Grade
        return result

obj = Point(22)
obj1 = Point(23)
print(obj + obj1)
```

```python
"""16. Create a Book class and overload the + operator. Concept 
Operator Overloading 
What to do 
Add the number of pages of two books. 
How to solve 
1. Create class Book.  
2. Store pages.  
3. Overload __add__().  
4. Return the total pages.  """

class Book:
    def __init__(self,page):
        self.page = page

    def __add__(self,other):
        total = self.page + other.page
        return total

obj1 = Book(44)
obj2 = Book(50)
print(obj1 + obj2)
```

```python
"""17. Create a Student class and overload the > operator. Concept 
Comparison Operator Overloading (__gt__) 
What to do 
Compare marks of two students. 
How to solve 
1. Create class Student.  
2. Store marks.  
3. Overload __gt__().  
4. Compare the marks.  
5. Return True or False.
"""
class Student:
    def __init__(self,marks):
        self.marks = marks

    def __gt__(self,other):
        result = self.marks > other.marks
        return result

obj1 = Student(89)
obj2 = Student(90)

print(obj1 > obj2)
```

```python
"""
18. Create a Rectangle class and overload the == operator.
Sumit Sir Notes 
Concept 
Equality Operator Overloading (__eq__) 
What to do 
Compare rectangles using their area. 
How to solve 
1. Create class Rectangle.  
2. Store length and width.  
3. Calculate area.  
4. Overload __eq__().  
5. Compare the areas.  

"""

class Rectangel:
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def __eq__(self,other):
        result = (self.length * self.width) == (other.length * other.width)
        return result


obj = Rectangel(3,2)
obj1 = Rectangel(3,5)

print(obj == obj1)

    
```

```python
"""
19. Create a Money class and overload the + and - operators. Concept 
Multiple Operator Overloading 
What to do 
Perform addition and subtraction of money amounts. How to solve 
1. Create class Money.  
2. Store the amount.  
3. Overload __add__().  
4. Overload __sub__().  
5. Return new Money objects after each operation.  
 

"""
class Money:
    def __init__(self,amount):
        self.amount = amount

    def __add__(self,other):
        return Money(self.amount + other.amount)

    def __sub__(self,other):
        return Money(self.amount - other.amount)

obj = Money(2333)
obj1 = Money(4433)

type(obj + obj1)
type(obj - obj1)
```
