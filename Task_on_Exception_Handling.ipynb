```python
class InvalidOperatorError(Exception):
    pass

def calculator(num1,num2,operator):
    if operator == '+':
        print("Addition is :",num1 + num2)
    elif operator == '-':
        print("Substraction is :",num1 - num2)
    elif operator == '*':
        print("Multiplication is :",num1 * num2)
    elif operator == '/':
        print("Division is :",num1 / num2)
        
    elif operator == '//':
        print("floor division is :",num1 // num2)
    elif operator == '**':
        print("Power is :",num1 ** num2)
    elif operator == '%':
        print("Remender is :",num1 % num2)
    else:
        raise InvalidOperatorError("Error: Invalid Operator you Enter")

try:

    n1 = int(input())
    n2 = int(input())
    op = input()
    calculator(n1,n2,op)

except InvalidOperatorError as e:
    print(e)

except ZeroDivisionError:
    print("Error: Cannot divide by zero")

except ValueError:
    print("Error: Please enter valid integer values")


```

## 28. Create a banking system with deposit and withdraw. Raise and handle custom exception  for insufficient balance. 

```python
class InsufficientBalanceError(Exception):
    pass
    
class Bank:
    def __init__(self,balance):
        self.__Balance = balance

    # get balance
    def get_balance(Balance):
        return self.__Balance

    # setter balance
    def set_balance(Balance,new_balance):
        if new_balance > 0:
            self.__Balance = new_balance
            print("New balance added successfully")
        else:
            print("New balance is less than 0")

    # method 1 : Deposit
    def deposit(self):
        amount = float(input("Enter the amount for deposit:"))
        if amount > 0 :
            self.__Balance += amount
            print(f"₹{amount} is added successfully and your balance is {self.__Balance}")
            
        else:
            print("Amount is less than 0")

    # method 2 : withdraw
    def withdraw(self):
        amount = float(input("Enter the amount for withdraw:"))
        if self.__Balance > amount:
            self.__Balance -= amount
            print(f"₹{amount} are withdrawn in your account")
            print(f"Your current Balance is ₹{self.__Balance}")
        else:
            raise InsufficientBalanceError(f"Error : Your Balance is Insufficienct\nBalace is ₹{self.__Balance}")

try:

    obj = Bank(2000)
    obj.deposit()
    obj.withdraw()

except InsufficientBalanceError as e:
    print(e)
    
```
