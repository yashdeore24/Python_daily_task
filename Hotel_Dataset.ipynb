## 2. Standardize all date formats into one consistent format (YYYY-MM-DD)


```python
df["Booking Date"] = pd.to_datetime(
    df['Booking Date'],
    errors = 'coerce' ,
    format = 'mixed',
    dayfirst = True
)
```

```python
df["Booking Date"] = pd.to_datetime(
    df['Booking Date'],
    errors = 'coerce' ,
    format = 'mixed',
    dayfirst = True
).dt.strftime('%Y-%m-%d')
df['Booking Date'].fillna('2025-05-17',inplace = True)
```

```python
df['Booking Date'].value_counts()
```

## 3. Handle completely missing Customer Name values

```python
df['Customer Name'].value_counts()
```

```python
df['Customer Name'].isna().sum()
```

```python
df['Customer Name'] = df['Customer Name'].replace(
    ['N/A', 'NA', 'n/a', ''],
    pd.NA
)
```

```python
df['Customer Name'].fillna('Unknown', inplace = True)
```

## 4. Fix inconsistent Gender values

```python
df['Gender'] = df['Gender'].str.strip().str.lower()
```

```python
df['Gender'].value_counts()
```

```python
df['Gender'].replace({
    'm':'Male',
    'f':'Female',
    'male':'Male',
    'female':'Female',
    'other' : 'Other'})
```

## 5. Clean and standardize City names 


```python
df['City'].value_counts()
```

```python
df['City'].replace({
    'delhi':'Delhi',
    'mumbai':'Mumbai',
    'MUMBAI':'Mumbai',
    'Bangalore':'Bengaluru',
    'Hydrabad':'Hyderabad',
    'bangalore':'Bengaluru'
},inplace =True)
```

## 6. Standardize Room Type categories 


```python
df['Room Type'].value_counts()
```

```python
df['Room Type']=df['Room Type'].replace({
    'delux' : 'Deluxe',
    'exec' :  'Executive',
    'suite': 'Suite',
    'standard' : 'Standard',
    'STANDARD' : 'Standard',
    'Unknown' : 'Deluxe'
})
```

```python
df['Room Type'].value_counts()
```

## 7. Clean Booking Channel values 

```python
df['Booking Channel'].value_counts()
```

```python
df['Booking Channel']=df['Booking Channel'].str.strip()
```

```python
df['Booking Channel']=df['Booking Channel'].replace({
    'ONLINE' : 'Online',
    'TravelAgent' :  'Travel Agent',
    'corp': 'Corporate',
    'walkin' : 'Walk-in',
    'online' : 'Online',
    'travel agent' : 'Travel Agent'
})
```

```python
df['Booking Channel'].value_counts()
```

## 8. Standardize Customer Segment 


```python
df['Customer Segment'].value_counts()
```

```python
df['Customer Segment'] = df['Customer Segment'].replace({
    'regular':'Regular',
    'VIP':'Premium',
    'REG':'Regular',
    'corporate':'Corporate',
    'premium':'Premium',
    'budget':'Budget'
})
```

```python
df['Customer Segment'].value_counts()
```

## 9. Fix Age column (invalid and non-numeric values) 


```python

df['Age']=df['Age'].str.strip()
df['Age'].value_counts()
```

```python
df.info()
```

```python
df['Age'] = df['Age'].replace({
    '25 years' : 25,
    '30 yrs' : 30,
    '45Y' : 45,
    'unknown' :
})
```

```python
df['Age'].value_counts()
```

```python

df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
```

```python
df.loc[(df['Age'] < 18) | (df['Age'] > 100), 'Age'] = np.nan
```

```python
df['Age'] = df['Age'].astype('Int64')
```
