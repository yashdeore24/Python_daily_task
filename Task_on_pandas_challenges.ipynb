## Challenge 1 – Complete Data Cleaning

```python
# import pandas library
import pandas as pd
# load data data frame
df = pd.read_csv("C:\\Users\\deore\\Downloads\\Task_dataset - Task_dataset.csv")
df = pd.DataFrame(df)
df.head(15)
```

```python
# display dataset information
df.info()
```

```python
# find null values 
df.isna()
```

```python
# remove duplicate rowes
df.drop_duplicates(inplace=True)
df
```

```python
# fill the median in null value  in age column
df.fillna({ 'Age': df['Age'].median(), 
            'Balance': df['Balance'].mean() 
          }, 
          inplace=True)
df.head(15)
```

```python
# rename the Target --> CustomerStatus and Anuualincome --> Income
df.rename(columns = { 'Target': 'CustomerStatus',
                      'AnnualIncome' : 'Income'
                    },
          inplace = True)
                      
```

```python
# print first 15 rows of the data frame
df.head(15)
```
