## Challenge 11 – Missing Value Report

```python
# import pandas library
import pandas as pd
# load data data frame
df11 = pd.read_csv("C:\\Users\\deore\\Downloads\\Task_dataset - Task_dataset.csv")
df11 = pd.DataFrame(df11)


```

```python
# total missing values

Total_Missing_values = df11.isna().sum()
Total_Missing_values
```

```python
# Numerical columns

numeric_cols = df11.select_dtypes(include = 'number').columns.tolist()
```

```python
# mean of balance
mean = df11['Balance'].mean()
```

```python
# median of Age
median = df11['Age'].median()
```

```python
# fill missing values 
df11.fillna({'Balance' : df11['Balance'].mean(), 'Age' : df11['Age'].median()},inplace =True )
```

```python
# display all information
print("Total missing values :",Total_Missing_values)
print("Numerical colums are:")
for i in numeric_cols:
    print(i,end =",")

print("\n")
print("Mean of Balance columne :",mean)
print("Median of Age columne :",median)

    
```
