## 11.Task

```python
rev = 0 
digit = 0
num = int(input("Enter any number"))

while num > 0:
    digit = num % 10
    rev = digit + rev * 10 
    num = num // 10

print(rev)
    
```

## 12.Task

```python
num = int(input("Enter any number: "))

prime = True
while True:
for j in range(2, num):
    if num % j == 0:
        prime = False
        break

if prime and num > 1:
    print("Prime Number")
else:
    print("Composite Number")
```

## 22.Task

```python
n = int(input("Enter the size of the square: "))

for i in range(1, n + 1):
    for j in range(1, n + 1):
        if i == 1 or i == n or j == 1 or j == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```
