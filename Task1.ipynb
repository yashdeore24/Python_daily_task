
# 9. Simple Calculator


```python
# Python
# Create a simple calculator that takes two numbers and one operator (+,-,*,/,//,%,**)
# from the user and shows the result.

num1 = float(input("Enter first number:"))
operater = str(input("Enter opration like + - * / // ** % :"))
num2 = float(input("Enter secound number:"))

if operater == "+":
    print("Addition is:",num1 + num2)

elif operater == "-":
    print("Addition is:",num1 - num2)

elif operater == "*":
    print("Multiplication is:",num1 * num2)

elif operater == "/":
    if num2 > 0 :
        print("division is:",num1 / num2)
    else :
        print("pleas enter number grater than zeor")

elif operater == "//":
    if   num2 > 0 :
        print("floor division  is:",num1 // num2)
    else:
        print("pleas enter number grater than zeor")

elif operater == "%":
    if   num2 > 0 :
        print("Remainder is:",num1 % num2)
    else:
        print("pleas enter number grater than zeor")
    

elif operater == "**":
    print("Multiplication is:",num1 ** num2)

else:
    print("Invalid operator")
```

# 10. Temperature Converter

```python
# Convert temperature from Celsius to Fahrenheit and vice versa.
# Formula: F = (C * 9/5) + 32

Celsius = int(input("Enter temperature in Celsius:"))

F = (Celsius * 9 / 5) + 32
print("Temperature in fahrenheit:" ,F)

C = (F - 32) * 5/9 
print("Temperature in celsius:",C)
```

# 12. Student Grade Calculator

```python

# Python
# Take marks of 5 subjects.
# Calculate percentage and print grade:
# >=90 → A+, >=80 → A, >=70 → B, >=60 → C, else → Fail

English = float(input("Enter the marks obtain in English:"))
Marathi = float(input("Enter the marks obtain in Marathi:"))
Hindi = float(input("Enter the marks obtain in Hindi:"))
Maths = float(input("Enter the marks obtain in Maths:"))
Science = float(input("Enter the marks obtain in science:"))

Total_Marks = English + Marathi + Hindi + Maths + Science
percentage = (Total_Marks / 500) * 100

print("Total Marks obtain in Exam:", Total_Marks)
print("Percentage obtain in Exam:", percentage)

if percentage >= 90:
    print("A+")

elif percentage >= 80:
    print("A")

elif percentage >= 70:
    print("B")

elif percentage >= 60:
    print("C")

else:
    print("Fail")
          
```
