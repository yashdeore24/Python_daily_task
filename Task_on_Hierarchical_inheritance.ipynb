```python
class Person:

    def __init__(self,name,age):
        self.Person_name = name
        self.Person_age = age

    

class Student(Person):

    def show(self):
        print("Student name:",self.Person_name)
        print("Student age:",self.Person_age)

    
    

class Teacher(Person):

    def show(self):
        print("Teacher name:",self.Person_name)
        print("Teacher age:",self.Person_age)

obj = Teacher("Rohit",22)
obj1 = Student("Yash",22)

obj.show()
obj1.show()
```

```python
"""
3. Vehicle → Car, Bike
Question
Create class Vehicle with method start(). Create subclasses Car and Bike.
Concept
• One parent method shared by multiple children.
How to Solve
1. Create Vehicle.
2. Add start().
3. Inherit it in Car.
4. Inherit it in Bike.
5. Call start() from both child objects."""

class Vehicle:

    def start(self):
        print("Car and Bike both are vehicle")

class Car(Vehicle):

    pass
    
class Bike(Vehicle):

    pass

obj = Bike()
obj1 = Car()

obj.start()
obj1.start()
    
```

```python
"""
4. Shape → Circle, Rectangle
Question
Define class Shape with method area(). Create subclasses Circle and Rectangle.
Concept
• Parent provides common method.
• Children may later override it.
How to Solve
1. Create Shape.
2. Add area().
3. Create Circle.
4. Create Rectangle.
5. Call area().
"""



class Shape:

    def __init__(self,name):
        self.name=name

    def area(self):
        print("Area of a Shape",self.name)
    

class Circle(Shape):

    pass
    
class Rectangle(Shape):

    pass

obj = Circle("Circle")
obj1 = Rectangle("Rectangle")

obj.area()
obj1.area()
```

```python
"""
5 Question
5. Employee → Manager, Developer
Sumit Sir Notes
Create class Employee with attributes emp_id and salary. Create subclasses Manager and
Developer.
Concept
• Attribute inheritance.
How to Solve
1. Create parent constructor.
2. Store emp_id and salary.
3. Create two child classes.
4. Show inherited attributes.
"""

class Employee:

    def __init__(self,emp_id,salary):
        self.Emp_id = emp_id
        self.Salary = salary
        
        

class Manager(Employee):

    def show(self):
        print("Manager id :",self.Emp_id)
        print("Manager salary :",self.Salary)
        
    

class Developer(Employee):
    
    def display(self):
        print("Developer id :",self.Emp_id)
        print("Developer salary :",self.Salary)

obj = Developer(12453,40000)
obj2 = Manager(12345,30000)

obj.display()
obj2.show()

        
```

```python
"""
7. Device → Mobile, Laptop
Question
Create class Device with method turn_on().
Concept
• Reusing common functionality.
How to Solve
1. Create parent.
2. Add turn_on().

Sumit Sir Notes
3. Create two children.
4. Demonstrate inheritance.

"""


class Device:

    def __init__(self,name):
        self.name = name

    def turn_on(self):
        print(f"Turn on the {self.name}")

class Mobile(Device):

    pass

class Laptop(Device):

    pass


obj = Mobile("Mobile")
obj1 = Laptop("Laptop")

obj.turn_on()
obj1.turn_on()



```

```python
"""
8. Sports → Cricket, Football
Question
Define class Sports with method play().
Concept
• Common method inherited.
How to Solve
1. Parent class.
2. Common method.
3. Two subclasses.
4. Call inherited method.
"""

class Sports:
    def __init__(self,name):
        self.name = name

    def play(self):
        print(f"play the {self.name}")

class Cricket(Sports):

    pass
    

class Football(Sports):

    pass

obj = Football("Football")
obj1 = Cricket("Cricket")

obj.play()
obj1.play()
```

```python
"""
9. Product → Electronics, Clothing
Question
Create class Product with attribute price.
Concept
• Common data shared by children.
How to Solve
1. Create constructor.
2. Store price.
3. Create child classes.
4. Display price.

"""


class Product:
    def __init__(self,price,pro):
        self.price=price
        self.pro = pro

    def display(self):
        print(f"Price of {self.pro} product is {self.price}")
        
        
class Electronics(Product):
    pass

    
class Clothing(Product):
    pass


obj = Clothing(1000,"Clothing")
obj1 = Electronics(1200,"Electronics")

obj.display()
obj1.display()
    
```

```python
"""
10. Book → Fiction, NonFiction
Question
Define class Book with attributes title and author.

Sumit Sir Notes
Concept
• Constructor inheritance.
How to Solve
1. Parent constructor.
2. Two child classes.
3. Access inherited attributes.

"""

class Book:
    def __init__(self,title,author):
        self.title = title
        self.author = author

    def show(self):
        print("Book title is :", self.title)
        print("Book author is :" , self.author)

class Fiction(Book):
    pass
    
class NonFiction(Book):
    pass 

obj = Fiction("The Inheritance Games"," Jennifer Lynn Barnes")
obj1 = NonFiction("Educated","Tara Westover")

obj.show()
obj1.show()
```
