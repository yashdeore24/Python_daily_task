```python
"""
12. Transport → Bus, Train 
Question 
Define class Transport with method move(). Concept 
• One behavior for many children.  
How to Solve 
1. Parent class.  
2. Add method.  
3. Create Bus.  
4. Create Train.  
5. Demonstrate inheritance. 

"""

class Transport:
    def __init__(self,way,v):
        self.way=way
        self.v = v

    def move(self):
        print(f"{self.v} is move on {self.way}")
        
class Bus(Transport):
    pass
class Train(Transport):
    pass

obj = Bus("Road","Bus")
obj1 = Train("Track","Train")

obj.move()
obj1.move()
```

## Hierarchical Inheritance with Constructors & Methods (13–22) 

```python
"""
13. Person Constructor 
Concept 
Using super() to call the parent constructor. 
How to Solve 
1. Create constructor in parent.  
2. Create child constructors.  
3. Use super().__init__().  
4. Add child-specific attribute.  
5. Display all data.

"""

class Person:

    def __init__(self,surname):
        self.Surname = surname

    def display(self):
        print(f"Name is {self.Name} {self.Surname}")
        

class Child1(Person):

    def __init__(self,surname,name):
        super().__init__(surname)
        self.Name = name
        
class Child2(Person):

    def __init__(self,surname,name):
        super().__init__(surname)
        self.Name = name
    
obj = Child1(surname = "Deore", name ="Yash")
obj1 = Child2(name = "Tanmay", surname ="Deore")

obj.display()
obj1.display()
```

```python
"""
14. Animal Sound Override 
Concept 
Method Overriding. 
How to Solve 
1. Parent has sound().  
2. Dog overrides.  
3. Cat overrides.  
4. Call method from each object.  

"""

class Animal: 
    def sound(self): 
        print("Animal makes a sound") 
        
class Dog(Animal):
    def sound(self): # overriding 
        print("Dog barks") 
        
class Cat(Animal): 
    def sound(self): # overriding 
        print("Cat meows") 
        
obj = Dog() 
obj1 = Cat()
obj.sound() 
obj1.sound()
```

```python
"""
15. Vehicle Constructor 
Concept 
Constructor inheritance. 
How to Solve 
1. Parent stores brand.  
2. Car adds doors. 
Sumit Sir Notes 
3. Bike adds engine_cc.  
4. Use super().  

"""

class Vehicle:

    def __init__(self, brand):
        self.brand = brand


class Car(Vehicle):

    def __init__(self, brand, doors):
        super().__init__(brand)
        self.doors = doors

    def display(self):
        print(f"Car Brand: {self.brand}")
        print(f"Doors: {self.doors}")


class Bike(Vehicle):

    def __init__(self, brand, engine_cc):
        super().__init__(brand)
        self.engine_cc = engine_cc

    def display(self):
        print(f"Bike Brand: {self.brand}")
        print(f"Engine: {self.engine_cc} cc")


car = Car("Toyota", 4)
bike = Bike("Honda", 150)

car.display()
print()
bike.display()
```
