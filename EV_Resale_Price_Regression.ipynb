# 5. Battery Data Imputation 

```python
# Battery capacity -> median
df['battery_capacity_kwh'].fillna(df['battery_capacity_kwh'].median(), inplace=True)

# Battery health -> median
df['battery_health_pct'].fillna(df['battery_health_pct'].median(), inplace=True)

# Range -> median within battery capacity groups
df['range_km'] = df['range_km'].fillna(
    df.groupby('battery_capacity_kwh')['range_km'].transform('median')
)

# Charging time -> median
df['charging_time_hr'].fillna(df['charging_time_hr'].median(), inplace=True)
```

# 6. Battery Health Outlier Investigation 


```python
# Basic checks
below_min = df[df['battery_health_pct'] < 50]
above_100 = df[df['battery_health_pct'] > 100]

# IQR method for extreme observations
Q1 = df['battery_health_pct'].quantile(0.25)
Q3 = df['battery_health_pct'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

extreme = df[(df['battery_health_pct'] < lower) | (df['battery_health_pct'] > upper)]

print('Below 50% :', len(below_min))
print('Above 100 :', len(above_100))
print('Extreme   :', len(extreme))
```

# 7. Range vs Battery Capacity 


```python
# Create efficiency feature
df['range_per_kwh'] = df['range_km'] / df['battery_capacity_kwh']

# IQR method to detect unusual efficiency
Q1 = df['range_per_kwh'].quantile(0.25)
Q3 = df['range_per_kwh'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

low_eff = df[df['range_per_kwh'] < lower]
high_eff = df[df['range_per_kwh'] > upper]

print('Unusually low efficiency :', len(low_eff))
print('Unusually high efficiency:', len(high_eff))

# Show suspicious vehicles
print(df.loc[(df['range_per_kwh'] < lower) | (df['range_per_kwh'] > upper),
             ['vehicle_id', 'range_km', 'battery_capacity_kwh', 'range_per_kwh']])
```

# 8. Driving Intensity Feature

```python
# Handle vehicle_age = 0
df['km_per_year'] = df['km_driven'] / df['vehicle_age'].replace(0, pd.NA)

# Identify unusually high annual driving (IQR method)
Q1 = df['km_per_year'].quantile(0.25)
Q3 = df['km_per_year'].quantile(0.75)
IQR = Q3 - Q1

upper = Q3 + 1.5 * IQR

high_driving = df[df['km_per_year'] > upper]

print('Vehicle age = 0 :', (df['vehicle_age'] == 0).sum())
print('Unusually high annual driving :', len(high_driving))

# Show suspicious vehicles
print(high_driving[['vehicle_id', 'km_driven', 'vehicle_age', 'km_per_year']])
```

# 9. Charging Efficiency 


```python
# Missing values
print('Missing range_km :', df['range_km'].isna().sum())
print('Missing charging_time_hr :', df['charging_time_hr'].isna().sum())

# Avoid division by zero
df['charging_efficiency'] = df['range_km'] / df['charging_time_hr'].replace(0, pd.NA)

# Missing values in new feature
print('Missing charging_efficiency :', df['charging_efficiency'].isna().sum())

# Extreme values using IQR
Q1 = df['charging_efficiency'].quantile(0.25)
Q3 = df['charging_efficiency'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

extreme = df[(df['charging_efficiency'] < lower) | (df['charging_efficiency'] > upper)]

print('Extreme charging_efficiency :', len(extreme))

# Show suspicious rows
print(extreme[['vehicle_id', 'range_km', 'charging_time_hr', 'charging_efficiency']])

# Prepare for regression
df['charging_efficiency'] = df['charging_efficiency'].clip(lower=lower, upper=upper)
```

```python
import pandas as pd
import numpy as np

# =========================================================
# STEP 1 — AUDIT
# =========================================================

print('Shape:', df.shape)
print('\nInfo:')
print(df.info())

print('\nMissing values:')
print(df.isna().sum())

print('\nDuplicate rows:', df.duplicated().sum())

cat_cols = ['brand','vehicle_type','fast_charging','city','service_history']
for c in cat_cols:
    print(f'\n{c} distribution:')
    print(df[c].value_counts(dropna=False).head(10))

# =========================================================
# Preserve original dataset
# =========================================================

df_raw = df.copy(deep=True)
df_clean = df.copy()

# =========================================================
# STEP 2 — CLEAN IDENTIFIERS
# =========================================================

dup_ids = df_clean['vehicle_id'].duplicated(keep=False)

# If exact duplicate rows exist, remove exact duplicates only
df_clean = df_clean.drop_duplicates()

# Keep vehicle_id for validation, remove later from X

# =========================================================
# STEP 3 — CLEAN DATES
# =========================================================

df_clean['listing_date'] = pd.to_datetime(df_clean['listing_date'], errors='coerce')

df_clean['listing_year'] = df_clean['listing_date'].dt.year
df_clean['listing_month'] = df_clean['listing_date'].dt.month
df_clean['listing_quarter'] = df_clean['listing_date'].dt.quarter

# Remove invalid future dates
df_clean = df_clean[df_clean['listing_date'] <= pd.Timestamp.today()]

# =========================================================
# STEP 4 — CLEAN NUMERICAL FEATURES
# =========================================================

num_cols = [
    'battery_capacity_kwh',
    'battery_health_pct',
    'range_km',
    'km_driven',
    'charging_time_hr',
    'owner_count'
]

# Battery capacity
df_clean['battery_capacity_kwh'] = df_clean['battery_capacity_kwh'].fillna(
    df_clean['battery_capacity_kwh'].median()
)

# Battery health
df_clean = df_clean[df_clean['battery_health_pct'].ge(0) | df_clean['battery_health_pct'].isna()]
df_clean['battery_health_pct'] = df_clean['battery_health_pct'].clip(upper=100)
df_clean['battery_health_pct'] = df_clean['battery_health_pct'].fillna(
    df_clean['battery_health_pct'].median()
)

# Range by battery group
df_clean['range_km'] = df_clean['range_km'].fillna(
    df_clean.groupby('battery_capacity_kwh')['range_km'].transform('median')
)
df_clean['range_km'] = df_clean['range_km'].fillna(df_clean['range_km'].median())

# km_driven
df_clean = df_clean[df_clean['km_driven'].ge(0) | df_clean['km_driven'].isna()]
df_clean['km_driven'] = df_clean['km_driven'].fillna(df_clean['km_driven'].median())

# charging_time_hr
df_clean = df_clean[df_clean['charging_time_hr'].gt(0) | df_clean['charging_time_hr'].isna()]
df_clean['charging_time_hr'] = df_clean['charging_time_hr'].fillna(
    df_clean['charging_time_hr'].median()
)

# owner_count
df_clean = df_clean[df_clean['owner_count'].between(1, 10) | df_clean['owner_count'].isna()]
df_clean['owner_count'] = df_clean['owner_count'].fillna(
    round(df_clean['owner_count'].median())
).astype(int)

# =========================================================
# STEP 5 — CLEAN CATEGORICAL FEATURES
# =========================================================

for c in ['brand','vehicle_type','city']:
    df_clean[c] = (
        df_clean[c]
        .astype(str)
        .str.strip()
        .str.title()
        .replace({'Nan': pd.NA})
    )

# fast_charging -> binary
df_clean['fast_charging'] = (
    df_clean['fast_charging']
    .astype(str)
    .str.strip()
    .str.title()
    .replace({'Nan': pd.NA})
)

df_clean['fast_charging_bin'] = df_clean['fast_charging'].map({'Yes':1, 'No':0})

# service_history
df_clean['service_history'] = (
    df_clean['service_history']
    .replace('Missing', pd.NA)
    .fillna('Unknown')
    .astype(str)
    .str.strip()
    .str.title()
)

# Fill remaining categorical missing values
for c in ['brand','vehicle_type','city']:
    df_clean[c] = df_clean[c].fillna('Unknown')

# =========================================================
# STEP 6 — FEATURE ENGINEERING
# =========================================================

df_clean['vehicle_age'] = df_clean['listing_year'] - df_clean['manufacture_year']

# Remove impossible ages
df_clean = df_clean[df_clean['vehicle_age'].ge(0)]

df_clean['range_per_kwh'] = (
    df_clean['range_km'] / df_clean['battery_capacity_kwh']
)

df_clean['km_per_year'] = (
    df_clean['km_driven'] / df_clean['vehicle_age'].replace(0, np.nan)
)

df_clean['charging_efficiency'] = (
    df_clean['range_km'] / df_clean['charging_time_hr']
)

df_clean['battery_usage_index'] = (
    (df_clean['battery_health_pct'] / 100) /
    (1 + df_clean['km_driven'] / 100000)
)

# Handle infinities
df_clean.replace([np.inf, -np.inf], np.nan, inplace=True)

# Fill engineered-feature missing values
for c in ['range_per_kwh','km_per_year','charging_efficiency','battery_usage_index']:
    df_clean[c] = df_clean[c].fillna(df_clean[c].median())

# =========================================================
# STEP 7 — LEAKAGE CHECK
# =========================================================

leakage_features = []

for col in df_clean.columns:
    if col != 'resale_price':
        if 'resale_price' in col.lower():
            leakage_features.append(col)

# Remove any leakage features if present
df_clean.drop(columns=leakage_features, inplace=True, errors='ignore')

print('\nLeakage features removed:', leakage_features)

# =========================================================
# STEP 8 — FINAL VALIDATION
# =========================================================

print('\nFinal duplicate vehicle IDs:',
      df_clean['vehicle_id'].duplicated().sum())

print('Invalid dates:',
      df_clean['listing_date'].isna().sum())

print('Remaining missing values:')
print(df_clean.isna().sum()[df_clean.isna().sum() > 0])

print('Impossible vehicle ages:',
      (df_clean['vehicle_age'] < 0).sum())

print('Impossible battery health values:',
      ((df_clean['battery_health_pct'] < 0) |
       (df_clean['battery_health_pct'] > 100)).sum())

# Ensure numeric dtypes
numeric_expected = [
    'battery_capacity_kwh','battery_health_pct','range_km',
    'km_driven','charging_time_hr','owner_count',
    'vehicle_age','range_per_kwh','km_per_year',
    'charging_efficiency','battery_usage_index',
    'listing_year','listing_month','listing_quarter',
    'fast_charging_bin'
]

df_clean[numeric_expected] = df_clean[numeric_expected].apply(
    pd.to_numeric, errors='coerce'
)

# Final missing check
assert df_clean[numeric_expected].isna().sum().sum() == 0, 'Unexpected numeric missing values remain'

# =========================================================
# STEP 9 — FINAL OUTPUT
# =========================================================

y = df_clean['resale_price']

X = df_clean.drop(columns=[
    'resale_price',
    'vehicle_id',
    'listing_date',
    'fast_charging'
])

print('\nX shape:', X.shape)
print('y shape:', y.shape)

print('\nFeature types:')
print('Numerical:',
      X.select_dtypes(include=['int64','float64']).columns.tolist())

print('\nCategorical:',
      X.select_dtypes(include=['object','category','bool']).columns.tolist())

print('\nDataset is ready for regression modeling.')
```
