## 20. Create multi-level inheritance: Person → Employee → Manager. 


```python
class Person:
    
    def __init__(self,name,**key):
        super().__init__(**key)
        self.__name = name

    def get_name(self):
        return self.__name 
        
class Employee(Person):
    
    def __init__(self,ID,**key):
        super().__init__(**key)
        self.__ID= ID
        
    def get_ID(self):
        return self.__ID
    
class Manager(Employee):
    
    def __init__(self,name,ID,team_size,**key):
        super().__init__(name =name , ID = ID )
        
        self.__team_size = team_size

    def get_team_size(self):
        return self.__team_size

    def display(self):
        print("Manager details:")
        print("Name is :",self.get_name())
        print("ID is   :",self.get_ID())
        print("Team size:",self.get_team_size())

obj = Manager("Yash",1245,4)
obj.display()
    
    
```

## 23. Create abstract class Shape with abstract method area(). Implement it in Circle and  Rectangle using super(). 


```python
from abc import ABC, abstractmethod
import math
class Shape(ABC):

    def __init__(self,name):
        self.name = name
    

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):

    def __init__(self,r):
        super().__init__("Circle")
        self.r = r

    def area(self):
        print(f"Area of {self.name} is :{math.pi * self.r **2}")


class Rectangle(Shape):
    
    def __init__(self,length,width):
        super().__init__("Rectangle")
        self.length = length
        self.width = width

    def area(self):
        print(f"Area of {self.name} is : {self.length * self.width}")

obj = Rectangle(2,4)
obj.area()
obj = Circle(5)
obj.area()


        
```

## 30. Create class ComplexNumber and overload both + and - operators.


```python
class ComplexNumber:

    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def __add__(self, other):
        return ComplexNumber(self.real + other.real, self.imag + other.imag)

    def __sub__(self, other):
        return ComplexNumber(self.real - other.real, self.imag - other.imag)

    def __str__(self):
        return f"{self.real} + {self.imag}i"


obj1 = ComplexNumber(3, 4)
obj2 = ComplexNumber(5, 2)

print(obj1 + obj2)
print(obj1 - obj2)
```
