### 3. Write a program to check whether a triangle is:
#### o Equilateral
#### o Isosceles
#### o Scalene



```python
s1 = float(input())
s2 = float(input())
s3 = float(input())

if s1 + s2 > s3 and s2 + s3 > s1 and s3 + s1 > s2:
    if s1 == s2 == s3 :
        print("Equilateral")
    elif s1 == s2 or s2 == s3 or s3 == s1:
        print("Isosceles")
    elif s1 != s2 != s3:
        print("Scalene")
else :
    print("Invalid Triangle:")
    
```

### 1. Write a program to create a simple ATM system:
o Check balance
o Withdraw money
o Deposit money
o Invalid option message

```python
Balence = 1000
Input = input("Enter what you want:\n For Check balance : C \n For Withdraw money: Withdraw\n For Deposit money: Deposit\n")

Input = Input.upper()

if Input == 'C':
    
    print(f"Your Balence is :₹{Balence}")
elif Input == "WITHDRAW":
    Amount = float(input("Enter the amount:"))
        
    if Amount <= Balence:
        Balence -= Amount
        print(f"Take the :₹{Amount}")
    else:
        print(f"Your balence is:{Balence}")
                    
elif Input == "DEPOSIT MONEY":
      Deposit = float(input("Enter the amount:")) 
      Balence += Deposit
      print(f"You deposited ₹{Deposit} and your Balence is ₹{Balence}")

else:
    print("Invalid otion message")
```

### 3. Write a program to determine the type of roots in a quadratic equation using discriminant.

```python

a = float(input("Enter the value of a: "))
b = float(input("Enter the value of b: "))
c = float(input("Enter the value of c: "))

if a == 0:
    print("This is not quadratic equation:")
else :
    d = b**2 - 4*a*c

    if d > 0:
        print("The equation has two distinct real roots.")
    elif d == 0:
        print("The equation has two equal real roots.")
    else:
        print("The equation has two complex (imaginary) roots.")
```
